# **Objective:**

• **Save and retrieve processed data efficiently inside Dataproc.**

• **Serve data in a structured way for analysis.**

• **Use Parquet, Hive, and CSV**


```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder\
.appName('Olist Ecommerce Optimization')\
.config('spark.executor.memory','6g')\
.config('spark.executor.cores','4')\
.config('spark.driver.memory','4g')\
.config('spark.MaxResultSize','2g')\
.config('spark.shuffle.partitions','64')\
.config('spark.default.parallelism','64')\
.config('spark.sql.adaptive.enabled','true')\
.config('spark.sql.adaptive.coalescePartition.enabled','true')\
.config('autoBroadcastJoinThreshold',20*1024*1024)\
.config('spark.sql.files.maxPartitionBytes','64MB')\
.config('spark.sql.files.openCostInBytes','2MB')\
.config('spark.memory.storagFraction',0.2)\
.config('spark.memory.fraction',0.8)\
.getOrCreate()
```

```python
!hadoop fs -mkdir /data/tem/
```

```python
!hadoop fs -ls  /data/
```

```python
full_orders_df=spark.read.parquet('/data/olist_processed/cleaned_data_parquet/')
```

```python
# Save as parquet in HDFS
full_orders_df.write.mode('overwrite').parquet('/data/serve')
```

```python
# save it as a parquet in a google cloud storage
full_orders_df.write.mode('overwrite').parquet('gs://dataproc-staging-us-east1-704297063092-k4nmd8kh/temp_data')
```

```python
# save it as table
full_orders_df.write.mode('overwrite').saveAsTable('cleaned_data')
```

```python
spark.sql('show tables')
```

```python
# save it as csv
full_orders_df.write.mode('overwrite').option('header','true').csv('/data/tem/')
```
